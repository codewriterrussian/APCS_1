# Python：矩陣操作與反向還原

這份筆記整理「**依照操作紀錄，把矩陣反向還原**」的做法。

主要會用到：

1. 二維串列（2D List）
2. `reversed()`
3. `[::-1]`
4. 巢狀迴圈
5. 矩陣上下翻轉
6. 矩陣旋轉
7. `zip(*matrix)` 的轉置技巧

---

## 核心觀念

如果矩陣依序做了：

```text
操作 A → 操作 B → 操作 C
```

要回到原本狀態，就必須反過來：

```text
C 的反操作 → B 的反操作 → A 的反操作
```

因此程式中會看到：

```python
for operation in reversed(operations):
```

這是整題最重要的觀念之一。


# 1. 讀取矩陣

輸入的第一行包含：

```text
R C M
```

其中：

- `R`：矩陣列數
- `C`：矩陣欄數
- `M`：操作次數

可以寫成：

```python
R, C, M = [int(x) for x in input().split()]
```

接著讀取 `R` 列矩陣資料。


````python
R, C, M = [int(x) for x in input().split()]

matrix = []

for _ in range(R):
    row = [int(x) for x in input().split()]
    matrix.append(row)

operations = [int(x) for x in input().split()]
````

matrix = [

1 2 3

matrix[row][column]

# 2. 為什麼要反向處理操作？

假設操作紀錄是：

```python
operations = [0, 1, 0]
```

原本執行順序為：

```text
0 → 1 → 0
```

還原時一定要從最後一個操作開始：

```text
0 → 1 → 0
↑       ↑
最後    最先
```

因此使用：

```python
reversed(operations)
```


````python
operations = [0, 1, 0]

for operation in reversed(operations):
    print(operation)
````

# 3. `[::-1]`：反轉串列

Python 中：

```python
lst[::-1]
```

可以把串列順序完全反過來。


````python
lst = [1, 2, 3, 4, 5, 6, 7, 8, 9]

print(lst[::-1])
````

如果套用在二維串列：

```python
matrix[::-1]
```

反轉的是「**列的順序**」。

例如：

```text
1 2 3
4 5 6
7 8 9
```

變成：

```text
7 8 9
4 5 6
1 2 3
```

因此：

```python
matrix = matrix[::-1]
```

可以完成**上下翻轉**。


````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix = matrix[::-1]

for row in matrix:
    print(*row)
````

# 4. 操作 `1`：上下翻轉

上下翻轉有一個很重要的特性：

> 做兩次上下翻轉，就會回到原本的矩陣。

因此它自己的反操作仍然是「上下翻轉」。

程式可以直接寫：

```python
if operation == 1:
    matrix = matrix[::-1]
```


# 5. 操作 `0`：反向旋轉

依照原 notebook 的還原邏輯，遇到操作 `0` 時，需要把矩陣做 **90° 逆時針旋轉**。

例如：

```text
原矩陣

1 2 3
4 5 6
```

逆時針旋轉後：

```text
3 6
2 5
1 4
```

注意：

- 原本是 `2 × 3`
- 旋轉後變成 `3 × 2`

所以旋轉後，矩陣的列數與欄數會交換。


## 方法一：使用巢狀迴圈

這種寫法比較長，但最適合第一次理解矩陣索引的變化。

```python
new_matrix = []

for i in range(len(matrix[0])):
    row = []

    for j in range(len(matrix)):
        row.append(matrix[j][-(i + 1)])

    new_matrix.append(row)

matrix = new_matrix
```

### 關鍵

```python
matrix[j][-(i + 1)]
```

會從每一列的右邊開始取元素。

因此可以依序建立逆時針旋轉後的新矩陣。


````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

new_matrix = []

for i in range(len(matrix[0])):
    row = []

    for j in range(len(matrix)):
        row.append(matrix[j][-(i + 1)])

    new_matrix.append(row)

matrix = new_matrix

for row in matrix:
    print(*row)
````

# 6. 使用 `zip(*matrix)` 轉置矩陣

Python 可以利用：

```python
zip(*matrix)
```

把矩陣的「列」與「欄」交換。

例如：

```text
1 2 3
4 5 6
```

轉置後：

```text
1 4
2 5
3 6
```


````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

transposed = [list(row) for row in zip(*matrix)]

for row in transposed:
    print(*row)
````

## 利用「轉置 + 上下反轉」完成逆時針旋轉

先做：

```python
[list(row) for row in zip(*matrix)]
```

得到轉置矩陣。

再使用：

```python
[::-1]
```

反轉列的順序。

因此可以寫成：

```python
matrix = [list(row) for row in zip(*matrix)][::-1]
```

這就是較精簡的 **90° 逆時針旋轉**。


````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix = [list(row) for row in zip(*matrix)][::-1]

for row in matrix:
    print(*row)
````

# 7. 完整解法：容易理解的版本

下面保留原 notebook 使用巢狀迴圈的思路，並整理變數名稱與排版。

```python
for operation in reversed(operations):
```

代表從最後一個操作開始逐一還原。


````python
R, C, M = [int(x) for x in input().split()]

matrix = []

for _ in range(R):
    row = [int(x) for x in input().split()]
    matrix.append(row)

operations = [int(x) for x in input().split()]


# 必須從最後一個操作開始還原
for operation in reversed(operations):

    # 操作 1：上下翻轉
    if operation == 1:
        matrix = matrix[::-1]

    # 操作 0：逆時針旋轉 90°
    else:
        new_matrix = []

        for i in range(len(matrix[0])):
            row = []

            for j in range(len(matrix)):
                row.append(matrix[j][-(i + 1)])

            new_matrix.append(row)

        matrix = new_matrix


# 輸出最後的矩陣大小
R = len(matrix)
C = len(matrix[0])

print(R, C)

# 輸出矩陣
for row in matrix:
    print(*row)
````

# 8. 完整解法：較精簡的 Python 寫法

理解前面的索引操作後，可以利用 Python 的切片與 `zip()` 簡化：

```python
if operation == 1:
    matrix = matrix[::-1]
else:
    matrix = [list(row) for row in zip(*matrix)][::-1]
```

考試時如果對 `zip(*matrix)` 還不熟，使用上一個巢狀迴圈版本會更容易除錯。


````python
R, C, M = [int(x) for x in input().split()]

matrix = []

for _ in range(R):
    matrix.append([int(x) for x in input().split()])

operations = [int(x) for x in input().split()]


for operation in reversed(operations):

    if operation == 1:
        # 上下翻轉
        matrix = matrix[::-1]

    else:
        # 逆時針旋轉 90°
        matrix = [list(row) for row in zip(*matrix)][::-1]


print(len(matrix), len(matrix[0]))

for row in matrix:
    print(*row)
````

# 9. 常見錯誤

## 錯誤 1：沒有反向處理操作

錯誤：

```python
for operation in operations:
```

還原矩陣時應該從最後一個操作開始：

```python
for operation in reversed(operations):
```

---

## 錯誤 2：旋轉後仍使用原本的 `R`、`C`

矩陣旋轉後：

```text
R × C
```

會變成：

```text
C × R
```

因此最後直接使用：

```python
len(matrix)
len(matrix[0])
```

最安全。

---

## 錯誤 3：混淆 `matrix[::-1]` 與每一列的反轉

```python
matrix[::-1]
```

是：

> 上下翻轉

而：

```python
[row[::-1] for row in matrix]
```

則是：

> 左右翻轉

兩者不同。


# 10. 重點整理

```text
reversed(operations)
    → 從最後一個操作開始還原

matrix[::-1]
    → 上下翻轉

row[::-1]
    → 單一列左右反轉

zip(*matrix)
    → 矩陣轉置

[list(row) for row in zip(*matrix)][::-1]
    → 逆時針旋轉 90°
```

解這類矩陣題時，可以先問自己：

> **原本做了什麼操作？它的反操作是什麼？**

再問：

> **多個操作還原時，應該從哪一個操作開始？**

答案是：

> **從最後一個操作開始。**
