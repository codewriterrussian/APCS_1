```python
R, C, k, m = [int(i) for i in input().split()]
city = []
for i in range(R):
    city.append([int(j) for j in input().split()])

for M in range(m):
    new_city = [row[:] for row in city]

    directions = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for i in range(R):
        for j in range(C):

            move = city[i][j] // k

            for di, dj in directions:
                ni = i + di
                nj = j + dj

                
                if 0 <= ni < R and 0 <= nj < C: 
                    if city[ni][nj] != -1 and city[i][j] != -1:
                        new_city[i][j] -= move
                        new_city[ni][nj] += move
    city = [row[:] for row in new_city]
    

bigest = -1
smallest = 99999999999
for i in range(R):
    for j in range(C):
        if city[i][j] > bigest:
            bigest = city[i][j]
        elif city[i][j] < smallest and city[i][j] != -1:
            smallest = city[i][j]

print(smallest)
print(bigest)
```

```python
n, m = map(int, input().split())

S = list(map(int, input().split()))
T = list(map(int, input().split()))

# 題目編號是 1 ~ n，轉成 Python 的 0 ~ n-1
order = [x - 1 for x in map(int, input().split())]

lose_count = [0] * n


while len(order) > 1:

    winners = []
    losers = []

    # 每次兩個人一組
    for i in range(0, len(order) - 1, 2):

        p = order[i]
        q = order[i + 1]

        # 一定要先保存「比賽前」的數值
        a = S[p]
        b = T[p]

        c = S[q]
        d = T[q]

        # p 獲勝
        if a * b >= c * d:

            # 勝利者 p
            S[p] = a + (c * d) // (2 * b)
            T[p] = b + (c * d) // (2 * a)

            # 失敗者 q
            S[q] = c + c // 2
            T[q] = d + d // 2

            lose_count[q] += 1

            winners.append(p)

            if lose_count[q] < m:
                losers.append(q)

        # q 獲勝
        else:

            # 勝利者 q
            S[q] = c + (a * b) // (2 * d)
            T[q] = d + (a * b) // (2 * c)

            # 失敗者 p
            S[p] = a + a // 2
            T[p] = b + b // 2

            lose_count[p] += 1

            winners.append(q)

            if lose_count[p] < m:
                losers.append(p)

    # 奇數人的最後一位直接晉級
    if len(order) % 2 == 1:
        winners.append(order[-1])

    # 下一 round
    order = winners + losers


# 轉回 1-based 編號
print(order[0] + 1)
```

```python
R, C, k, m = map(int, input().split())

matrix = []

for _ in range(R):

    row = list(map(int, input().split()))
    matrix.append(row)


directions = [
    (-1, 0),
    (1, 0),
    (0, -1),
    (0, 1)
]


for _ in range(m):

    # 明天的人口
    new_matrix = [row[:] for row in matrix]

    for r in range(R):

        for c in range(C):

            # -1 不是城市
            if matrix[r][c] == -1:
                continue

            move = matrix[r][c] // k

            # 看四個方向
            for dr, dc in directions:

                nr = r + dr
                nc = c + dc

                # 地圖範圍內，而且是城市
                if (
                    0 <= nr < R
                    and 0 <= nc < C
                    and matrix[nr][nc] != -1
                ):

                    new_matrix[r][c] -= move
                    new_matrix[nr][nc] += move

    # 明天變成新的今天
    matrix = new_matrix


# 找所有真正的城市
population = []

for r in range(R):
    for c in range(C):
        if matrix[r][c] != -1:
            population.append(matrix[r][c])


print(min(population))
print(max(population))
```
