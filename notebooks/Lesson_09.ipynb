## 題目 1 — 字串置換／重建

### 學生解法


```python
K, Q, R = [int(i) for i in input().split()]
S = input()

ans = []

for i in range(Q):
    lst = [int(k) for k in input().split()]
    new_S = [""] * K

    for j in range(K):
        # 將原本位置 j 的字元
        # 移動到 lst[j] 指定的位置。
        new_S[lst[j] - 1] = S[j]

    S = "".join(new_S)
    ans.append(S)

for i in range(R):
    ans_for_print = ""

    for j in range(Q):
        ans_for_print += ans[j][i]

    print(ans_for_print)
```

### 較精簡的比較版本

演算法與原本相同，但變數名稱與輸入處理方式做了簡化，讓程式更容易閱讀。


```python
K, Q, R = map(int, input().split())
s = input().strip()

history = []

for _ in range(Q):
    permutation = list(map(int, input().split()))
    new_s = [""] * K

    for j in range(K):
        new_s[permutation[j] - 1] = s[j]

    s = "".join(new_s)
    history.append(s)

for r in range(R):
    result = ""

    for q in range(Q):
        result += history[q][r]

    print(result)
```

### Python 補充 — `str.strip()`

簡單複習 `strip()`：它可以移除字串開頭與結尾指定的字元；若沒有指定字元，預設會移除前後空白。


```python
print("   A B C ".strip())
print("###hello###".strip("#"))
```

## 題目 2 — 字串轉換

### 學生嘗試版本

這個版本保留下來作為除錯與比較用途。特別要注意：

- 計算 `"1"` 數量時所比較的資料型態。
- 控制最後重新排列字串的條件。


```python
m, n = [int(i) for i in input().split()]

lst = []
for i in range(m):
    e = input()
    lst.append(e[::-1])

lst = lst[::-1]
T = input()

for i in range(m):
    S = ""

    for j in range(n):
        if lst[i][j] == "1":
            S += T[-j - 1]
        else:
            S = S[::-1]
            S += T[-j - 1]
            S = S[::-1]

    T = S
    count = 0

    for k in lst[i]:
        if k == 1:  # 學生版本：保留型別不一致的問題，方便檢討。
            count += 1

    if count % 2 != 1:
        note_front = ""
        note_middle = ""
        note_back = ""
        note = ""

        if len(T) % 2 != 0:
            for l in range(len(T)):
                if l <= (len(T) - 1) / 2:
                    note_front += T[l]
                elif l == (len(T) + 1) / 2:
                    note_middle += T[l]
                else:
                    note_back += T[l]

            note += note_front
            note += note_middle
            note += note_back
            T = note

        else:
            for l in range(len(T)):
                if l <= len(T) / 2:
                    note_front += T[l]
                else:
                    note_back += T[l]

            note += note_front
            note += note_back
            T = note

print(T)
```

### 修正版

這個版本修正了 `"1"` 的比較方式，並在 `"1"` 的數量為奇數時，依照原本預期的方式將字串的兩半重新排列。


```python
m, n = [int(i) for i in input().split()]

lst = []
for i in range(m):
    e = input()
    lst.append(e[::-1])

lst = lst[::-1]
T = input()

for i in range(m):
    S = ""

    for j in range(n):
        if lst[i][j] == "1":
            S += T[-j - 1]
        else:
            S = S[::-1]
            S += T[-j - 1]
            S = S[::-1]

    T = S
    count = 0

    for k in lst[i]:
        if k == "1":
            count += 1

    if count % 2 != 0:
        note_front = ""
        note_middle = ""
        note_back = ""
        note = ""

        if n % 2 != 0:
            for l in range(n):
                if l <= (n - 3) / 2:
                    note_front += T[l]
                elif l == (n - 1) / 2:
                    note_middle += T[l]
                else:
                    note_back += T[l]

            note += note_back
            note += note_middle
            note += note_front
            T = note

        else:
            for l in range(n):
                if l <= (n - 2) / 2:
                    note_front += T[l]
                else:
                    note_back += T[l]

            note += note_back
            note += note_front
            T = note

print(T)
```
