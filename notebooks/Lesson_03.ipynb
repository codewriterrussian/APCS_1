## 第 1 題：b964. 成績指標

### 題目要做什麼？

給定 `n` 位學生的成績，我們要完成三件事：

1. **把所有成績由小到大輸出**。
2. 找出 **不及格成績中最高的分數**（也就是最接近 60、但仍小於 60 的分數）。
3. 找出 **及格成績中最低的分數**（也就是最接近 60、且大於等於 60 的分數）。

如果沒有任何不及格成績，輸出：

```text
best case
```

如果沒有任何及格成績，輸出：

```text
worst case
```

> 這題很適合練習：串列、排序、迴圈、條件判斷，以及「找最大值 / 最小值」的思考方式。


### 範例輸入

```text
10
0 11 22 33 55 66 77 99 88 44
```

第一行：

```text
10
```

表示有 `10` 個成績。

第二行：

```text
0 11 22 33 55 66 77 99 88 44
```

是 10 位學生的成績。

---

### 手動分析

排序後：

```text
0 11 22 33 44 55 66 77 88 99
```

其中：

- 小於 `60` 的成績：`0, 11, 22, 33, 44, 55`
- 最高的不及格成績：`55`
- 大於等於 `60` 的成績：`66, 77, 88, 99`
- 最低的及格成績：`66`

所以輸出應為：

```text
0 11 22 33 44 55 66 77 88 99
55
66
```


### 解題想法

我們可以把解法拆成兩個階段。

#### 第一步：排序

使用：

```python
p.sort()
```

排序完成後，成績會由小到大排列。

#### 第二步：找兩個關鍵成績

我們需要：

- `h`：目前找到的 **最高不及格成績**
- `l`：目前找到的 **最低及格成績**

因為合法成績範圍通常在 `0 ~ 100`，所以可以先使用不可能出現的值作為「尚未找到」的標記：

```python
h = -1
l = 101
```

接著逐一檢查所有成績：

```python
if p[i] < 60:
    h = p[i]
```

因為串列已經排序，所以每遇到一個新的不及格成績，它都會比前一個更大，因此最後留下的就是最高不及格分數。

而及格部分：

```python
elif p[i] >= 60 and p[i] < l:
    l = p[i]
```

只要找到比目前 `l` 更小的及格分數，就更新 `l`。


### 程式逐段解釋

#### 1. 讀入學生人數

```python
n = int(input())
```

`input()` 讀到的是字串，所以要用 `int()` 轉成整數。

#### 2. 讀入所有成績

```python
p = [int(i) for i in input().split()]
```

拆開來看：

```python
input()
```

先讀入整行，例如：

```text
0 11 22 33 55 66 77 99 88 44
```

接著：

```python
.split()
```

會得到字串串列：

```python
['0', '11', '22', '33', '55', '66', '77', '99', '88', '44']
```

最後：

```python
[int(i) for i in ...]
```

把每一個字串轉成整數。

#### 3. 排序

```python
p.sort()
```

注意：`sort()` 會直接修改原本的串列。

#### 4. 輸出排序結果

```python
for i in range(n - 1):
    print(p[i], end=" ")
print(p[-1])
```

這樣寫是為了避免最後一個數字後面多一個空白。

`p[-1]` 代表串列最後一個元素。


````python
n = int(input())
p = [int(i) for i in input().split()]

# 由小到大排序
p.sort()

# 輸出排序後的成績
for i in range(n - 1):
    print(p[i], end=" ")
print(p[-1])

# h：最高不及格分數
# l：最低及格分數
# -1 和 101 都是「尚未找到」的標記值
h = -1
l = 101

for i in range(n):
    if p[i] < 60:
        # 因為已排序，所以後面遇到的不及格分數會更大
        h = p[i]
    elif p[i] >= 60 and p[i] < l:
        # 找到目前最低的及格分數
        l = p[i]

# 沒有不及格成績
if h == -1:
    print("best case")
else:
    print(h)

# 沒有及格成績
if l == 101:
    print("worst case")
else:
    print(l)
````

### 常見錯誤

#### 錯誤 1：忘記把輸入轉成整數

錯誤：

```python
p = input().split()
```

這樣得到的是字串。字串排序和數字排序可能不同，例如：

```text
'100' < '20'
```

在字串比較中可能成立，因此一定要轉成 `int`。

---

#### 錯誤 2：把 60 算成不及格

錯誤：

```python
if score <= 60:
```

正確應為：

```python
if score < 60:
```

因為 `60` 分已經及格。

---

#### 錯誤 3：`best case` / `worst case` 判斷顛倒

記法：

- 全部都及格 → 沒有不及格成績 → `best case`
- 全部都不及格 → 沒有及格成績 → `worst case`

---

#### 錯誤 4：只找第一個不及格成績

題目要的是 **最高的不及格成績**，不是任意一個不及格成績。

## 第 2 題：原 notebook 未附題目名稱

### 這段程式在做什麼？

第一行讀入兩個整數：

```python
a, b = [int(i) for i in input().split()]
```

接著讀入：

```python
n = int(input())
```

表示後面會處理 `n` 組資料。

每一組資料都會讀成串列 `m`：

```python
m = [int(i) for i in input().split()]
```

程式接著計算兩個值：

- `A`：`a` 出現時 `+1`，`-a` 出現時 `-1`
- `B`：`b` 出現時 `+1`，`-b` 出現時 `-1`

最後如果：

```python
A != 0 and B != 0
```

就把：

```python
total += 1
```

也就是說，這組資料會被計入答案。

最後輸出符合條件的資料組數：

```python
print(total)
```


### 為什麼要同時檢查正數和負數？

程式中有四種情況：

```python
if m[i] == a:
    A += 1
elif m[i] == b:
    B += 1
elif m[i] == -a:
    A -= 1
elif m[i] == -b:
    B -= 1
```

因此，對 `a` 而言：

- 出現 `a` → `A + 1`
- 出現 `-a` → `A - 1`

例如：

```text
a = 3
資料中有：3 3 -3
```

那麼：

```text
A = 1 + 1 - 1 = 1
```

所以：

```python
A != 0
```

成立。

但是如果資料是：

```text
3 -3
```

則：

```text
A = 1 - 1 = 0
```

此時：

```python
A != 0
```

不成立。

`B` 的計算方式完全相同。


### 變數整理

| 變數 | 用途 |
|---|---|
| `a` | 第一個指定整數 |
| `b` | 第二個指定整數 |
| `n` | 要處理的資料組數 |
| `m` | 目前這一組資料 |
| `A` | `a` 與 `-a` 抵消後的計數 |
| `B` | `b` 與 `-b` 抵消後的計數 |
| `total` | 最後符合條件的資料組數 |
| `i` | 走訪目前資料串列的索引 |
| `j` | 第幾組資料 |


````python
a, b = [int(i) for i in input().split()]
n = int(input())

total = 0

# 一共處理 n 組資料
for j in range(n):
    A = 0
    B = 0

    # 讀入目前這一組資料
    m = [int(i) for i in input().split()]

    # 逐一檢查資料中的每個數字
    for i in range(len(m)):
        if m[i] == a:
            A += 1
        elif m[i] == b:
            B += 1
        elif m[i] == -a:
            A -= 1
        elif m[i] == -b:
            B -= 1

    # a 與 b 對應的最後計數都不能是 0
    if A != 0 and B != 0:
        total += 1

print(total)
````

### 第 2 題常見閱讀重點

#### 1. `A`、`B` 每一組都要歸零

注意它們放在：

```python
for j in range(n):
```

裡面。

所以每開始處理新的一組資料，就重新計算。

如果把 `A = 0`、`B = 0` 放到外面，前一組的結果就會影響下一組。

---

#### 2. `elif` 表示一次只會執行一個分支

例如某個值符合：

```python
m[i] == a
```

執行 `A += 1` 後，就不會再檢查同一輪後面的 `elif`。

---

#### 3. `A != 0 and B != 0`

`and` 的意思是：**兩個條件都必須成立**。

只有：

```text
A 非 0
而且
B 非 0
```

這一組才會被加入 `total`。
