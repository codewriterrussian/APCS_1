## 學習方式
先用紙筆 trace 一輪，再寫函式，最後執行後方 `check(...)`。所有函式需 `return` 題目要求的結果，並且**不修改傳入的 list**。學生版 `pass` 是待作答區；檢查失敗表示尚未完成或答案有錯。

````python
import copy

def check(fn, cases):
    """比對回傳值，並檢查是否意外修改呼叫端資料。"""
    for i, (args, expected) in enumerate(cases, 1):
        inputs = copy.deepcopy(args)
        untouched = copy.deepcopy(inputs)
        actual = fn(*inputs)
        assert actual == expected, (f"Case {i} FAILED: expected {expected!r}, got {actual!r}")
        assert inputs == untouched, f"Case {i} FAILED: 不能修改傳入的資料"
    print(f"PASS — {fn.__name__}: {len(cases)} cases")
````
