```python
n, D = [int(i) for i in input().split()]

lst = [int(i) for i in input().split()]

stock = 1
x = lst[0]
Profit = -x
for i in range(1, n):
    if stock == 1:
        y = lst[i]
        if y >= x + D:
            Profit += y
            x = y
            stock = 0
    else:
        y = lst[i]
        if y <= x - D:
            Profit -= y
            x = y
            stock = 1

if stock == 1:
    Profit += x

print(Profit)
```

Mine

```python
n, D = map(int, input().split())
a = list(map(int, input().split()))

holding = True
price = a[0]
profit = 0

for current in a[1:]:
    if holding:
        if current >= price + D:
            profit += current - price 
            price = current
            holding = False
    else:
        if current <= price - D:
            price = current
            holding = True

print(profit)
```

```python
text = input()

string = ""
longest_string = ""

long = 1
longest = 1

for i in range(len(text) - 1):

    if ord(text[i]) == ord(text[i + 1]) - 1:
        long += 1
        string += text[i]

    else:

        if longest <= long:
            string += text[i]
            longest = long
            longest_string = string

        long = 1
        string = ""

if long >= longest:
    string += text[-1]
    longest_string = string
    longest = long

if longest == 1:
    longest_string = text[-1]

print(longest, longest_string)
```
