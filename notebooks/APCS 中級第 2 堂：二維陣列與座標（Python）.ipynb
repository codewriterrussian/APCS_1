````python
# 檢查工具（學生不要修改）
import copy

def _check(fn, cases):
    for index, (arguments, expected) in enumerate(cases, 1):
        original = copy.deepcopy(arguments)
        actual = fn(*copy.deepcopy(arguments))
        assert actual == expected, f"第 {index} 組：預期 {expected!r}，實際 {actual!r}"
        assert arguments == original, "檢查器的測資不應被改動"
    print(f"✅ 通過 {len(cases)} 組測試")
````

# 整合練習：原題的輸入／輸出格式

下面兩段程式可以用來理解 ZeroJudge 的資料輸入輸出，學生版請自己填 TODO。實際提交時，**只複製對應題目的完整程式區塊**，不要連同 Notebook 的其他 cell 一起提交。

- [o077 電子畫布](https://zerojudge.tw/ShowProblem?problemid=o077)：讀 `H W N`、N 筆 `(r,c,t,x)`；輸出 H 行、每行 W 個數字。
- [k732 特殊位置](https://zerojudge.tw/ShowProblem?problemid=k732)：讀 `n m`、n 行數字；先輸出特殊位置數量，再逐行輸出 `r c`（0-based）。

**本堂目標是理解，不要求兩題都立刻拿到 AC。**

## o077：完整輸入輸出模板

````python
def solve_o077(data):
    it = iter(map(int, data.split()))
    h, w, n = next(it), next(it), next(it)
    operations = [(next(it), next(it), next(it), next(it)) for _ in range(n)]
    canvas = q9_paint_many(h, w, operations)  # 請先完成 Q9
    return "\n".join(" ".join(map(str, row)) for row in canvas)

sample = "3 3 2\n1 1 1 4\n1 1 0 3"
print(solve_o077(sample))
# 預期：0 4 0 / 4 7 4 / 0 4 0（分三行）
````

## k732：完整輸入輸出模板

````python
def solve_k732(data):
    it = iter(map(int, data.split()))
    h, w = next(it), next(it)
    grid = [[next(it) for _ in range(w)] for _ in range(h)]
    ans = q12_special_positions(grid)  # 請先完成 Q12
    lines = [str(len(ans))]
    lines += [f"{r} {c}" for r, c in ans]
    return "\n".join(lines)

sample = "2 3\n5 2 3\n4 5 6"
print(solve_k732(sample))
# 預期：2 / 0 0 / 1 1（分三行）
````
