## 1. Q13：找出三個數值的中間值

函式：

```python
q13(a, b, c)
```

會比較 `a`、`b`、`c` 三個數值的大小關係。

程式利用連鎖比較：

```python
a <= b <= c
```

判斷某個數值是否位在另外兩個數值之間。


````python
# q13
def q13(a, b, c):
    if (a <= b <= c) or (c <= b <= a):
        return b
    elif (b <= a <= c) or (c <= a <= b):
        return a
    else:
        return c
````

### 判斷方式

如果：

```text
a <= b <= c
```

或：

```text
c <= b <= a
```

代表 `b` 位在另外兩個值之間，因此回傳 `b`。

接著檢查 `a` 是否位在 `b` 與 `c` 之間。

如果前面的情況都不成立，則回傳 `c`。

這種寫法也可以處理兩個數值相等的情況，因為使用的是 `<=`。


## 2. 算術運算子的計算順序

下面的程式直接計算一個包含：

- `+`
- `*`
- `/`

的算術式。


````python
print(12 + 2 * 24 / 3 * 4 + 4)
````

### 運算順序

在這個算式中：

```text
12 + 2 * 24 / 3 * 4 + 4
```

`*` 和 `/` 的優先順序高於 `+`。

因此可以先看成：

```text
12 + (2 * 24 / 3 * 4) + 4
```

而同一優先順序的 `*` 與 `/` 會由左到右計算：

```text
2 * 24
↓
48

48 / 3
↓
16.0

16.0 * 4
↓
64.0
```

最後再做加法：

```text
12 + 64.0 + 4
```

得到：

```text
80.0
```

因為 `/` 在 Python 中會產生浮點數結果，所以最後輸出也是浮點數。


## 3. 重點整理

### 連鎖比較

```python
a <= b <= c
```

等價於同時檢查：

```python
a <= b
b <= c
```

### 運算子優先順序

這份筆記中的算式可以先記住：

```text
*  /
先於
+  -
```

如果運算子優先順序相同，通常依照其規則由左到右進行計算。

遇到較複雜的算式時，也可以加入括號讓計算順序更清楚。
